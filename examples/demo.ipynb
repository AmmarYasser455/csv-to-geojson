# csv-to-geojson demo

Convert `examples/cities.csv` (eight world cities) to GeoJSON with the Python API.
Run from the repository root. On Colab, first run `!pip install git+https://github.com/AmmarYasser455/csv-to-geojson` and `!git clone https://github.com/AmmarYasser455/csv-to-geojson`, then run `%cd csv-to-geojson` in its own cell (a `cd` inside `!` does not persist).

```python
from csv_to_geojson import convert_file

result = convert_file("examples/cities.csv")
print(result.lat_column, result.lon_column, len(result.geojson["features"]), "features")
```

GeoJSON coordinates are `[longitude, latitude]`. Cairo is at 31.2°E, 30.0°N:

```python
for f in result.geojson["features"]:
    lon, lat = f["geometry"]["coordinates"]
    print(f"{f['properties']['city']:15} lon={lon:9.4f} lat={lat:8.4f}")
cairo = result.geojson["features"][0]["geometry"]["coordinates"]
assert cairo == [31.2357, 30.0444]
```

Rows that cannot be parsed are skipped and reported, not silently dropped:

```python
from csv_to_geojson import convert

messy = convert("name;lat;lon\nCairo;30,0444;31,2357\nNowhere;95;10\nOops;abc;10\n")
print(len(messy.geojson["features"]), "parsed")
for e in messy.errors:
    print("line", e.row, "-", e.reason)
assert len(messy.errors) == 2
```
